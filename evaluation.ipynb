```python
import pandas as pd
import re
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
#from args import *
import seaborn as sns
import matplotlib.pyplot as plt

pd.set_option("display.max_rows", None)
```

```python
def extract_prediction(response):
    try:
        prediction = re.search("\[(.*?)\]", response).group(1).replace('-', " ")
        #print(prediction)
        return prediction
    except:
        # print(response)
        # print("#" * 20)
        pass



def compute_accuracy(prediction_list, gold_list):
    correct = 0
    for prediction, gold in zip(prediction_list, gold_list):
        if prediction == gold:
            correct = correct + 1
    accuracy = correct / len(gold_list)
    return accuracy
```

```python
def Get_Prediction(df, prompt, version, df_classification_report):
    prediction_list = []
    for response in df["agg_response"]:
        #print(response)
        prediction = extract_prediction(response)
        prediction_list.append(prediction)

    df["Prediction"]= prediction_list
    #print(prediction_list)

    gold_list = df["label"].to_list()
    
    acc = compute_accuracy(prediction_list, gold_list)
    print(f"Accuracy: {acc: .3f}")

    target_names = ["TRUE", "HALF TRUE", "FALSE"]
    label_map = {"TRUE": 0, "HALF TRUE": 1, "FALSE": 2}
    labels = [label_map[e] for e in gold_list]
    predictions = [label_map[e] for e in prediction_list]
    print("Classification Report")
    print("=" * 60)
    report= classification_report(labels, predictions,
                                  target_names=target_names, digits=4,output_dict=True)
    #print(report)
    df_report = pd.DataFrame(report).transpose()
    df_report["prompt"]= prompt
    df_report["version"]= version
    df_classification_report= pd.concat([df_classification_report, df_report], axis=0)
    #print(df_classification_report)
    #csv_name= "./Results/evaluate
    #df_classification_report.to_csv(csv_name)
    print(confusion_matrix(labels, predictions))
    return df_classification_report
```

```python
df_classification_report= pd.DataFrame(columns=["prompt","version","precision", "recall" ,"f1-score",  "support" ])
```

```python
fewshot= pd.read_json("./PoliResult/out/aggregate/final0_start_150_sample_llama-8b_aggregate_fewshot_V1.0.json")
df_classification_report= Get_Prediction(fewshot, "fewshot", "V1.0", df_classification_report)
print(df_classification_report)
df_classification_report.drop(['accuracy', 'macro avg', 'weighted avg'], axis=0, inplace=True)
```

```python
import math
label_value={'TRUE':0, 'HALF TRUE':1, 'FALSE':2}

prompt_list= ["direct", "fewshot", "novice_questions", "journalist_questions", "no_questions"]
version_list= ['V1.0']

df_RMSE= pd.DataFrame(columns=["prompt","version","RMSE"])
rmse_out=[]
for ver in version_list:
    for prompt in prompt_list:
        file_name= "./PoliResult/out/aggregate/final0_start_150_sample_llama-8b_aggregate_" + prompt + "_" +ver + ".json"
        result= pd.read_json(file_name)
        #print(result.columns)
        prediction_list = []
        for response in result["agg_response"]:
            prediction = extract_prediction(response)
            prediction_list.append(prediction)
        result['prediction']= prediction_list
        
        #print(result)
        loss=0
        for i in range(len(result)):
            label = result.loc[i, 'label']
            prediction= result.loc[i, 'prediction']
            loss= loss + ((label_value[label]- label_value[prediction])**2)
            
        rmse= math.sqrt(loss/len(result))
        rmse_out.append({"prompt": prompt, "version": ver, "RMSE": rmse})
```
